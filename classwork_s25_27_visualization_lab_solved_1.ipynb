### Heatmap

```python
# Drop NAs and calculate the correlation matrix for numeric columns
numeric_penguins = penguins.select_dtypes(include=[np.number]).dropna()
corr_matrix = numeric_penguins.corr()

# Create the heatmap correlogram
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm_r", fmt=".2f")
plt.show()
```
